```python
# Step 6: Multi-Lingual Intent Prediction Benchmark (English, Hindi, Gujarati, Hinglish)
test_benchmarks = [
    ("Create file sem9_report.txt in D drive", "English CRUD"),
    ("Ek navin file banavo notes.txt ma hello", "Gujarati CRUD"),
    ("File data.json delete karo", "Hindi CRUD"),
    ("Open D drive", "Drive Controller"),
    ("What time is it right now?", "Temporal Intelligence"),
    ("Calculate 125 * 8", "Math Computation"),
    ("Tame kon cho?", "Gujarati Identity"),
    ("Who is Albert Einstein?", "World Knowledge"),
    ("What is my university?", "Episodic Memory Recall"),
    ("Remember that my project score is 100", "Memory Teaching")
]

print("=" * 75)
print(f"{'QUERY':<42} | {'PREDICTED INTENT':<22} | {'CONF':<6}")
print("=" * 75)
for q, category in test_benchmarks:
    intent, conf = mm.predict(q)
    print(f"{q:<42} | {intent:<22} | {int(conf*100)}%")
print("=" * 75)
```

```python
# Step 7: Demonstrate Persistent Memory & Unknown Query Handling
# 1. Recalls known taught facts
# 2. Teaches new dynamic memory facts
# 3. Recalls newly taught facts with 100% accuracy
# 4. Gracefully declines unknown questions with 'I am sorry, I do not know about this message'
from modules.nlp_engine import process_command

demo_scenarios = [
    "What is my university?",
    "What is my secret security key?",
    "Remember that my thesis topic is Deep Multimodal Transformers",
    "What is my thesis topic?",
    "What is randomnonsense999xyz?"
]

for prompt in demo_scenarios:
    resp, intent, meta = process_command(prompt)
    print(f"\n💬 Query : \"{prompt}\"")
    print(f"🎯 Intent: [{intent.upper()}]")
    print(f"⚡ Answer:\n{resp}")
    print("-" * 65)
```
